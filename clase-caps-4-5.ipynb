## Parte 4 — Trazar la trayectoria (capítulo 4)

La tortuga es un cursor que se mueve por un lienzo dejando rastro: `forward(n)` avanza n píxeles, `left(a)` y `right(a)` giran a grados. Con `make_turtle()` aparece el lienzo.

```python

# primer paseo: un cuadrado de lado 100, a pelo
t = make_turtle()
forward(50); left(90)
forward(50); left(90)
forward(50); left(90)
forward(50)
```

Cuatro líneas iguales repetidas... huele a función y a bucle, como en el capítulo 3.

```python
# mejor: encapsular el cuadrado
t = make_turtle()
for i in range(4):
    forward(70)
    left(90)
```

```python
def cuadrado():
    for i in range(4):
        forward(50)
        left(90)

t = make_turtle()
cuadrado()
```

Pero un cuadrado que solo sabe hacer lados de 100 no sirve para mucho. **Generalizar** es añadir un parámetro:

```python
def cuadrado(longitud):
    for i in range(4):
        forward(longitud)
        left(90)

t = make_turtle()
cuadrado(15)
cuadrado(21)
cuadrado(27)
cuadrado(33)
```

Y ahora el paso interesante del capítulo: un cuadrado es un polígono de 4 lados. En vez de otra función para pentágonos, otra para hexágonos..., una sola **interfaz** general: `poligono(n, longitud)`. El ángulo de giro de un polígono regular es 360/n — la figura sale sola de la geometría.

```python
def poligono(n, longitud):
    angulo = 360 / n
    for i in range(n):
        forward(longitud)
        left(angulo)

t = make_turtle(width= 300, height= 300)
poligono(5, 60)    # pentágono
poligono(8, 50)    # octógono
poligono(4, 30)
```

Un círculo no es más que un polígono con muchos lados:

```python
def circulo(radio):
    # longitud de cada lado para que el perímetro sea ~2*pi*r
    n = 60
    longitud = 2 * math.pi * radio / n
    poligono(n, longitud)

import math
t = make_turtle(width=400, height=500)
circulo(50)
circulo(80)
circulo(100)
```

Lo que acabamos de hacer, con nombre propio, es el corazón del capítulo 4:

- **Encapsulación**: meter código en una función (`cuadrado`).
- **Generalización**: añadir parámetros (`longitud`, luego `n`).
- **Diseño de interfaz**: `poligono(n, longitud)` es fácil de usar y difícil de usar mal.
- **Refactorización**: `circulo` no copia código, *reutiliza* `poligono`.

Falta un detalle de buena educación: el **docstring**, un string justo debajo del `def` que explica qué hace la función. Aparece cuando alguien hace `help(poligono)`.

```python
def poligono(n, longitud):
    """Dibuja un polígono regular de n lados.

    n: número de lados
    longitud: longitud de cada lado, en píxeles
    """
    angulo = 360 / n
    for i in range(n):
        forward(longitud)
        left(angulo)

help(poligono)
poligono(7, 90)
```

### if, else, elif

La **ejecución condicional**: el bloque indentado solo corre si la condición es `True`.

```python
def comprobar_combustible(litros):
    if litros > 50:
        print("Combustible suficiente")
    else:
        print("¡Aviso! Queda poco combustible")

comprobar_combustible(60)
comprobar_combustible(20)
```

```python
# con más de dos opciones: condicionales encadenados con elif
def estado_mision(litros):
    if litros > 70:
        print("Todo nominal")
    elif litros > 30:
        print("Ahorra combustible")
    else:
        print("Modo emergencia")

estado_mision(80)
estado_mision(45)
estado_mision(10)
```

Los `elif` se evalúan en orden y Python se queda con el **primero** que es `True`. Por eso el orden importa — a veces mucho:

```python
# ¿qué pasa si ponemos las condiciones al revés?
def estado_mision_mal(litros):
    if litros > 30:
        print("Ahorra combustible")
    elif litros > 70:
        print("Todo nominal")   # esta rama no se alcanza nunca
    else:
        print("Modo emergencia")

estado_mision_mal(80)   # debería decir "Todo nominal"...
```

### Condicionales anidados

Un `if` dentro de otro `if`. Funciona, pero se lee peor cuanto más niveles; normalmente un `and` o un `elif` lo deja más claro.

```python
def diagnostico(litros, oxigeno):
    if litros > 30:
        if oxigeno > 90:
            print("Todo ok")
        else:
            print("Revisa el oxígeno")
    else:
        print("Revisa el combustible")

diagnostico(60, 95)
diagnostico(60, 40)
diagnostico(10, 95)
```

### Recursión: la función que se llama a sí misma

Aquí viene la idea más bonita del capítulo. Una función puede llamarse a sí misma, siempre que haya un **caso base** que corte la cadena. La cuenta atrás del despegue es el ejemplo perfecto:

```python
def cuenta_atras(n):
    if n == 0:
        print("¡Despegue!")
    else:
        print(n)
        cuenta_atras(n - 1)

cuenta_atras(5)
```

Mirad la pila mentalmente: `cuenta_atras(5)` llama a `cuenta_atras(4)`, que llama a `cuenta_atras(3)`... hasta llegar a `0`, el caso base, y entonces cada llamada pendiente termina en orden inverso. Es el mismo diagrama de pila del capítulo 3, pero con la misma función apilada muchas veces.

Otro clásico: el factorial. `n! = n * (n-1)!`, con `0! = 1` como caso base.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

factorial(5)
```

```python
# esto va a fallar, y es instructivo: recursión sin caso base
def cuenta_atras_rota(n):
    print(n)
    cuenta_atras_rota(n - 1)

cuenta_atras_rota(3)   # RecursionError: Python pone un límite y para la sangría
```

`RecursionError: maximum recursion depth exceeded`. Sin caso base (o con un caso base que nunca se alcanza) la pila crece sin parar; Python corta al llegar al límite. Es la versión moderna de las torres de pilotes infinitas.
